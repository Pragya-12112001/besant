```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
marks = np.array([75, 80, 65, 90, 85])
print(marks)
```

```python
res = marks+5
print(res)
```

```python
print(np.sum(marks))
print(np.min(marks))
print(np.max(marks))
print(np.mean(marks))
print(np.average(marks))
```

```python
sales = np.array([50000, 80000, 120000, 90000, 150000])

print('average', np.average(sales))
print('min', np.min(sales))
print('max', np.max(sales))
print('increased', sales * 1.1)
print('sales avove 1 lakh', sales[sales>100000])
```

```python
data = {
  'name': ['ajay', 'bob', 'seri', 'jhon'],
  'department': ['cs', 'it', 'ece', 'cs'],
  'marks': [78, 76, 89, 87]
}

df = pd.DataFrame(data)
print('students detail: ', df)

# mean:
print('average marks: ', df['marks'].mean())
print('highest marks: ', df['marks'].max())
```

```python
print('students scoring above 80')
# print(df[df['marks']>80])
df[df['marks']>80]
```

```python
months = [
    'jan', 'feb', 'mar', 'apr', 'may', 'jun',
    'jul', 'aug', 'sep', 'oct', 'nov', 'dec'
]

sales = [
    28000, 32000, 35000, 31000,
    42000, 39000, 45000, 48000,
    44000, 55000, 62000, 70000
]

```

```python
plt.plot(months, sales)

plt.title('monthly sales')
plt.xlabel('months')
plt.ylabel('sales')

plt.show()
```

```python
plt.bar(months, sales)

plt.title('monthly sales')
plt.xlabel('months')
plt.ylabel('sales')

plt.show()
```
