# Exercise 1: Bill calculator
Write calculate_bill(price, quantity) to return the 
total cost.
Add a default discount parameter for 10%
Then:
• Call it for at least three items.
• For the third item, make the discount 5%
• Display the final bill amount

```python
def cal_bill(price, quant, discount = 10):
  total = price * quant
  return total - (total * discount/100)

print(cal_bill(1200, 4))
print(cal_bill(7500, 5))
print(cal_bill(3400, 7, discount=5))
```

# Exercise 3: Product Details
Write a function show_product(**details) that accepts and 
displays product information.
Requirements:
• Accept the product name and price as keyword 
arguments.
• Display each detail on a separate line

```python
def show_products(**details):
  for key, value in details.items():
    print(f'{key} : {value}')

show_products(name="Laptop", price=50000, brand="HP")
```

# Exercise 6: Scope
Create a Python program to calculate a shopping bill 
using global, local, and enclosing variables.
Requirements:
• Declare a global constant GST_RATE = 18.
• Create an outer function calculate_bill() with a 
variable price = 2000.
• Inside it, create an inner function show_bill() that 
calculates GST and final price
• Print the values of all variables from the appropriate 
functions.
• Print the original price again from the outer function.

```python

rate = 18  # Global variable (constant by convention)


def calculate_bill():
    price = 2000  # Enclosing variable for show_bill()

    def show_bill():
        gst = price * rate / 100  # Local variable
        final_price = price + gst     # Local variable

        print("GST Rate:", rate)
        print("Original Price:", price)
        print("GST Amount:", gst)
        print("Final Price:", final_price)

    show_bill()

    # Accessing the original price in the outer function
    print("Original Price from outer function:", price)


calculate_bill()
```
