 total prices

```python
prices = [120, 80, 150]
total = 0

for i in prices:
  total += i

print(total)
```

skip failed marks

```python
marks = [35, 72, 48, 91, 65]

for i in marks:
  if i<50:
    continue
  print(i)
```

find an item and report if not found

```python
fruits = ['apple', 'mango', 'banana']
target = 'hello'

for i in fruits:
  if i==target:
    print(i, 'target found')
    break
else: 
  print('target not found')
```

running total add till 5

```python
total = 0
n=1

while n<6:
  total += n
  n += 1
print(total)  
```
